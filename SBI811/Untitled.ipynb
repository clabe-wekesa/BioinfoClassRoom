```python
#!/usr/bin/env python3

from html import escape
from reportlab.lib import colors
from reportlab.lib.enums import TA_CENTER
from reportlab.lib.pagesizes import A4
from reportlab.lib.styles import ParagraphStyle, getSampleStyleSheet
from reportlab.lib.units import cm
from reportlab.platypus import (
    SimpleDocTemplate,
    Paragraph,
    Spacer,
    PageBreak,
    Table,
    TableStyle,
    Preformatted,
    KeepTogether,
    ListFlowable,
    ListItem,
)

OUTPUT_FILE = "Lecture_2_Git_GitHub_Project_Organization.pdf"

NAVY = colors.HexColor("#17365D")
BLUE = colors.HexColor("#1F4E79")
CODE_BACKGROUND = colors.HexColor("#EAF4FA")
CODE_TEXT = colors.HexColor("#124E78")
GREEN_BACKGROUND = colors.HexColor("#EDF7ED")
LIGHT_GREY = colors.HexColor("#F3F5F7")
BORDER = colors.HexColor("#B8CBD9")


# ---------------------------------------------------------------------
# Document setup
# ---------------------------------------------------------------------

document = SimpleDocTemplate(
    OUTPUT_FILE,
    pagesize=A4,
    rightMargin=1.7 * cm,
    leftMargin=1.7 * cm,
    topMargin=1.8 * cm,
    bottomMargin=1.7 * cm,
    title="Lecture 2: Git, GitHub and Project Organization",
    author="SBI 811",
)

base_styles = getSampleStyleSheet()

title_style = ParagraphStyle(
    "LectureTitle",
    parent=base_styles["Title"],
    fontName="Helvetica-Bold",
    fontSize=22,
    leading=27,
    alignment=TA_CENTER,
    textColor=NAVY,
    spaceAfter=7,
)

subtitle_style = ParagraphStyle(
    "LectureSubtitle",
    parent=base_styles["Normal"],
    fontName="Helvetica-Bold",
    fontSize=12,
    leading=16,
    alignment=TA_CENTER,
    textColor=BLUE,
    spaceAfter=16,
)

heading_style = ParagraphStyle(
    "SectionHeading",
    parent=base_styles["Heading1"],
    fontName="Helvetica-Bold",
    fontSize=15,
    leading=19,
    textColor=NAVY,
    spaceBefore=13,
    spaceAfter=7,
    keepWithNext=True,
)

subheading_style = ParagraphStyle(
    "Subheading",
    parent=base_styles["Heading2"],
    fontName="Helvetica-Bold",
    fontSize=11.5,
    leading=15,
    textColor=BLUE,
    spaceBefore=9,
    spaceAfter=5,
    keepWithNext=True,
)

body_style = ParagraphStyle(
    "Body",
    parent=base_styles["BodyText"],
    fontName="Helvetica",
    fontSize=9.8,
    leading=14,
    textColor=colors.HexColor("#222222"),
    spaceAfter=6,
)

small_style = ParagraphStyle(
    "Small",
    parent=body_style,
    fontSize=8.8,
    leading=12,
)

note_style = ParagraphStyle(
    "Note",
    parent=body_style,
    fontName="Helvetica-Bold",
    textColor=colors.HexColor("#7A3E00"),
)

code_style = ParagraphStyle(
    "Code",
    fontName="Courier",
    fontSize=8.3,
    leading=11,
    textColor=CODE_TEXT,
    leftIndent=0,
    rightIndent=0,
    spaceBefore=0,
    spaceAfter=0,
)

table_heading_style = ParagraphStyle(
    "TableHeading",
    parent=body_style,
    fontName="Helvetica-Bold",
    textColor=colors.white,
    alignment=TA_CENTER,
)

table_text_style = ParagraphStyle(
    "TableText",
    parent=body_style,
    fontSize=8.8,
    leading=12,
    spaceAfter=0,
)


# ---------------------------------------------------------------------
# Formatting helpers
# ---------------------------------------------------------------------

def p(text, style=body_style):
    """Create a paragraph."""
    return Paragraph(text, style)


def heading(text):
    return Paragraph(text, heading_style)


def subheading(text):
    return Paragraph(text, subheading_style)


def code_box(text, background=CODE_BACKGROUND):
    """Place terminal commands or file content in a coloured box."""
    formatted = Preformatted(escape(text.strip()), code_style)

    table = Table([[formatted]], colWidths=[16.2 * cm])
    table.setStyle(
        TableStyle(
            [
                ("BACKGROUND", (0, 0), (-1, -1), background),
                ("BOX", (0, 0), (-1, -1), 0.7, BORDER),
                ("LEFTPADDING", (0, 0), (-1, -1), 9),
                ("RIGHTPADDING", (0, 0), (-1, -1), 9),
                ("TOPPADDING", (0, 0), (-1, -1), 7),
                ("BOTTOMPADDING", (0, 0), (-1, -1), 7),
            ]
        )
    )
    return table


def bullet_list(items):
    return ListFlowable(
        [
            ListItem(
                Paragraph(item, body_style),
                leftIndent=12,
            )
            for item in items
        ],
        bulletType="bullet",
        start="circle",
        leftIndent=18,
        bulletFontName="Helvetica",
        bulletFontSize=7,
        spaceAfter=6,
    )


def numbered_list(items):
    return ListFlowable(
        [
            ListItem(
                Paragraph(item, body_style),
                leftIndent=14,
            )
            for item in items
        ],
        bulletType="1",
        start="1",
        leftIndent=22,
        spaceAfter=6,
    )


def information_table(headers, rows, widths):
    data = [
        [Paragraph(header, table_heading_style) for header in headers]
    ]

    for row in rows:
        data.append(
            [Paragraph(str(cell), table_text_style) for cell in row]
        )

    table = Table(data, colWidths=widths, repeatRows=1)
    table.setStyle(
        TableStyle(
            [
                ("BACKGROUND", (0, 0), (-1, 0), NAVY),
                ("GRID", (0, 0), (-1, -1), 0.5, BORDER),
                ("VALIGN", (0, 0), (-1, -1), "TOP"),
                ("ROWBACKGROUNDS", (0, 1), (-1, -1),
                 [colors.white, LIGHT_GREY]),
                ("LEFTPADDING", (0, 0), (-1, -1), 6),
                ("RIGHTPADDING", (0, 0), (-1, -1), 6),
                ("TOPPADDING", (0, 0), (-1, -1), 6),
                ("BOTTOMPADDING", (0, 0), (-1, -1), 6),
            ]
        )
    )
    return table


def footer(canvas, doc):
    canvas.saveState()
    canvas.setStrokeColor(colors.HexColor("#D5DEE5"))
    canvas.setLineWidth(0.5)
    canvas.line(
        document.leftMargin,
        1.25 * cm,
        A4[0] - document.rightMargin,
        1.25 * cm,
    )
    canvas.setFont("Helvetica", 8)
    canvas.setFillColor(colors.HexColor("#52616B"))
    canvas.drawString(
        document.leftMargin,
        0.8 * cm,
        "SBI 811 — Lecture 2",
    )
    canvas.drawRightString(
        A4[0] - document.rightMargin,
        0.8 * cm,
        f"Page {doc.page}",
    )
    canvas.restoreState()


# ---------------------------------------------------------------------
# Lecture content
# ---------------------------------------------------------------------

story = []

story.extend(
    [
        Paragraph("LECTURE 2", title_style),
        Paragraph(
            "Git, GitHub and Project Organization",
            subtitle_style,
        ),
        Paragraph(
            "<b>SBI 811: Computer Programming in Bioinformatics</b>",
            subtitle_style,
        ),
        heading("Learning outcomes"),
        p("By the end of this session, students should be able to:"),
        bullet_list(
            [
                "explain the difference between Git and GitHub;",
                "distinguish local and remote repositories;",
                "obtain course materials using <b>clone</b> and "
                "<b>pull</b>;",
                "organize files within a project;",
                "use <b>status</b>, <b>add</b>, <b>commit</b>, "
                "<b>pull</b> and <b>push</b>;",
                "explain the purpose of <b>README.md</b> and "
                "<b>.gitignore</b>;",
                "submit a first tracked exercise through a personal "
                "GitHub repository.",
            ]
        ),
        heading("1. Why do we need Git?"),
        p(
            "Suppose you are developing a program and save several "
            "versions using different names:"
        ),
        code_box(
            """
analysis.py
analysis_new.py
analysis_final.py
analysis_final2.py
analysis_final_correct.py
"""
        ),
        p(
            "After some time, it becomes difficult to know which file is "
            "current, what changed or whether an earlier version can be "
            "recovered."
        ),
        p(
            "<b>Git</b> is a version-control system. It records what "
            "changed, when it changed, who made the change and why the "
            "change was made. It also allows earlier versions of a project "
            "to be recovered."
        ),
        p(
            "Git is useful for scripts, notebooks, documentation and small "
            "project files. Very large sequencing and alignment files, such "
            "as FASTQ and BAM files, should normally be stored elsewhere."
        ),
        heading("2. Git and GitHub"),
        information_table(
            ["Git", "GitHub"],
            [
                [
                    "A program installed on the computer.",
                    "An online platform for hosting repositories.",
                ],
                [
                    "Tracks changes and creates commits.",
                    "Stores and shares repositories.",
                ],
                [
                    "Can work without an internet connection.",
                    "Requires internet for online operations.",
                ],
                [
                    "Records changes locally.",
                    "Receives commits through <b>git push</b>.",
                ],
            ],
            [8.1 * cm, 8.1 * cm],
        ),
        Spacer(1, 7),
        p(
            "Git and GitHub work together, but they are not the same "
            "thing. Git performs the version tracking; GitHub provides "
            "online storage and collaboration."
        ),
    ]
)

story.append(PageBreak())

story.extend(
    [
        heading("3. What is a repository?"),
        p(
            "A <b>repository</b>, usually shortened to <b>repo</b>, is a "
            "project directory whose files and history are tracked by Git."
        ),
        bullet_list(
            [
                "<b>Local repository:</b> the copy stored on your computer.",
                "<b>Remote repository:</b> the copy stored online on GitHub.",
            ]
        ),
        code_box(
            """
GitHub repository  -- git pull -->  Your computer
Your computer      -- git push -->  GitHub repository
"""
        ),
        heading("4. Project organization"),
        p(
            "A project should have a clear structure. A student's exercise "
            "repository may be organized as follows:"
        ),
        code_box(
            """
SBI811-Exercises/
|-- README.md
|-- .gitignore
|-- exercise_01/
|   `-- answers.md
|-- exercise_02/
`-- exercise_03/
"""
        ),
        p("A bioinformatics project may use the following structure:"),
        code_box(
            """
gene-analysis/
|-- README.md
|-- .gitignore
|-- data/
|   `-- sequences.fasta
|-- scripts/
|   `-- calculate_gc.py
|-- results/
|   `-- gc_results.tsv
`-- notebooks/
    `-- exploration.ipynb
"""
        ),
        subheading("Good project-organization practices"),
        bullet_list(
            [
                "Use meaningful file and directory names.",
                "Keep related files in the same directory.",
                "Separate data, programs, notebooks and results.",
                "Avoid names such as <b>new.py</b>, <b>final2.py</b> and "
                "<b>untitled.txt</b>.",
                "Include a README that introduces the project.",
            ]
        ),
        heading("5. Obtaining the course materials"),
        p(
            "<b>The lecturer's repository is used to distribute course "
            "materials. Students should not submit work or push changes "
            "to this repository.</b>",
            note_style,
        ),
        code_box(
            """
https://github.com/clabe-wekesa/BioinfoClassRoom
""",
            GREEN_BACKGROUND,
        ),
        p(
            "<b>git clone</b> downloads an existing repository to the "
            "computer. Begin by creating a directory for bioinformatics "
            "work:"
        ),
        code_box(
            """
cd ~
mkdir -p Bioinformatics
cd Bioinformatics
git clone https://github.com/clabe-wekesa/BioinfoClassRoom.git
cd BioinfoClassRoom
pwd
ls
"""
        ),
        p(
            "Cloning is normally performed only once. Do not clone the "
            "repository every time new materials are added."
        ),
    ]
)

story.append(PageBreak())

story.extend(
    [
        heading("6. Updating the course materials"),
        p(
            "After cloning the class repository once, use "
            "<b>git pull</b> to obtain newly uploaded materials:"
        ),
        code_box(
            """
cd ~/Bioinformatics/BioinfoClassRoom
git pull
ls
"""
        ),
        information_table(
            ["Command", "When it is used"],
            [
                [
                    "<b>git clone URL</b>",
                    "Download a repository for the first time.",
                ],
                [
                    "<b>git pull</b>",
                    "Update a repository that has already been cloned.",
                ],
            ],
            [5 * cm, 11.2 * cm],
        ),
        heading("7. Creating a personal exercise repository"),
        p(
            "Each student should create a separate GitHub repository. This "
            "protects the class repository while allowing every student to "
            "practise committing and pushing."
        ),
        p("Use the following settings on GitHub:"),
        information_table(
            ["Setting", "Value"],
            [
                ["Repository name", "SBI811-Exercises"],
                ["Description", "Practical exercises for SBI 811"],
                ["Visibility", "Public"],
                ["Add README", "Yes"],
                ["Add .gitignore", "Python"],
            ],
            [5 * cm, 11.2 * cm],
        ),
        Spacer(1, 7),
        p("The repository address will have the following form:"),
        code_box(
            """
https://github.com/STUDENT-USERNAME/SBI811-Exercises
"""
        ),
        subheading("Configure your Git identity"),
        p(
            "Before making the first commit, tell Git the name and email "
            "that should be attached to your work:"
        ),
        code_box(
            """
git config --global user.name "Your Full Name"
git config --global user.email "your-email@example.com"

git config --global user.name
git config --global user.email
"""
        ),
        p(
            "This is normally done only once on a computer. Preferably use "
            "an email address connected to your GitHub account."
        ),
        subheading("Clone your personal repository"),
        code_box(
            """
cd ~/Bioinformatics
git clone https://github.com/STUDENT-USERNAME/SBI811-Exercises.git
cd SBI811-Exercises
pwd
ls -la
"""
        ),
        p("You now have two separate repositories:"),
        code_box(
            """
~/Bioinformatics/
|-- BioinfoClassRoom/     # Course materials
`-- SBI811-Exercises/     # Your exercises
""",
            GREEN_BACKGROUND,
        ),
    ]
)

story.append(PageBreak())

story.extend(
    [
        heading("8. The Git workflow"),
        code_box(
            """
Working directory --git add--> Staging area
Staging area      --git commit--> Local repository
Local repository --git push--> GitHub
""",
            GREEN_BACKGROUND,
        ),
        bullet_list(
            [
                "<b>Working directory:</b> files currently being created "
                "or edited.",
                "<b>Staging area:</b> changes selected for the next commit.",
                "<b>Local repository:</b> checkpoints recorded on the "
                "computer.",
                "<b>Remote repository:</b> the online copy on GitHub.",
            ]
        ),
        heading("9. Essential Git commands"),
        information_table(
            ["Command", "Purpose"],
            [
                ["git status", "Show the current condition of the repository."],
                ["git add FILE", "Select a change for the next commit."],
                [
                    'git commit -m "message"',
                    "Record the staged changes locally.",
                ],
                ["git pull", "Obtain and incorporate remote changes."],
                ["git push", "Upload local commits to GitHub."],
            ],
            [6 * cm, 10.2 * cm],
        ),
        subheading("Checking the repository"),
        code_box("git status"),
        p(
            "Git may report files as <b>untracked</b>, <b>modified</b> or "
            "<b>staged</b>. An untracked file exists in the project but "
            "Git has not begun tracking it."
        ),
        subheading("Selecting changes"),
        code_box(
            """
git add exercise_01/answers.md   # Add one file
git add exercise_01/             # Add one directory
git add .                        # Add all current changes
"""
        ),
        p(
            "Run <b>git status</b> before using <b>git add .</b> so that "
            "unwanted files are not accidentally included."
        ),
        subheading("Recording a checkpoint"),
        code_box(
            """
git commit -m "Complete exercise 01"
"""
        ),
        p(
            "A commit records a checkpoint in the local repository. It "
            "does not automatically upload the work to GitHub."
        ),
        information_table(
            ["Good commit messages", "Poor commit messages"],
            [
                ["Complete exercise 01", "Changes"],
                ["Add FASTA sequence example", "Work"],
                ["Correct Git exercise answers", "Final"],
            ],
            [8.1 * cm, 8.1 * cm],
        ),
        subheading("Downloading and uploading changes"),
        code_box(
            """
git pull
git push
"""
        ),
        p(
            "<b>git pull</b> obtains changes from GitHub. "
            "<b>git push</b> uploads committed changes. A file that has "
            "not been committed cannot be pushed."
        ),
    ]
)

story.append(PageBreak())

story.extend(
    [
        heading("10. The README.md file"),
        p(
            "GitHub displays README.md on the repository's main page. It "
            "should identify the project and explain what it contains."
        ),
        code_box(
            """
# SBI 811 Exercises

This repository contains my practical exercises for SBI 811:
Computer Programming in Bioinformatics.

## Student

- Name: Your name
- Registration number: Your registration number
- GitHub username: Your username

## Completed exercises

- Exercise 01: Git and GitHub
"""
        ),
        p(
            "The .md extension means Markdown. A hash creates a heading, "
            "double asterisks create bold text, single asterisks create "
            "italic text and a hyphen creates a list item."
        ),
        heading("11. The .gitignore file"),
        p(
            "A .gitignore file tells Git which files or directories it "
            "should not track."
        ),
        code_box(
            """
# Python-generated files
__pycache__/
*.pyc

# Temporary files
*.tmp
*.log

# Editor files
.vscode/

# Large bioinformatics files
*.fastq
*.fastq.gz
*.bam

# Secret information
.env
config.json
"""
        ),
        p(
            "<b>Never upload passwords, API keys, access tokens, private "
            "patient information or confidential research data.</b>",
            note_style,
        ),
    ]
)

story.append(PageBreak())

story.extend(
    [
        heading("12. First tracked exercise"),
        p(
            "Complete this exercise inside your personal "
            "<b>SBI811-Exercises</b> repository."
        ),
        subheading("Required structure"),
        code_box(
            """
SBI811-Exercises/
|-- README.md
|-- .gitignore
`-- exercise_01/
    |-- answers.md
    `-- sequence.fasta
"""
        ),
        subheading("Questions for answers.md"),
        numbered_list(
            [
                "What is Git?",
                "What is GitHub?",
                "What is a repository?",
                "What is the difference between <b>git clone</b> and "
                "<b>git pull</b>?",
                "What is the difference between <b>git commit</b> and "
                "<b>git push</b>?",
                "What information does <b>git status</b> provide?",
                "Why should a project contain a README?",
                "Give four examples of files that should not be uploaded "
                "to GitHub.",
            ]
        ),
        subheading("Create the exercise files"),
        code_box(
            """
mkdir -p exercise_01
nano exercise_01/answers.md
nano exercise_01/sequence.fasta
"""
        ),
        p("Enter the following record in sequence.fasta:"),
        code_box(
            """
>sample_sequence
ATGCGTACGTTAGCTAGCTAGCTA
"""
        ),
        subheading("Track and submit the exercise"),
        code_box(
            """
git status
git add exercise_01/
git status
git commit -m "Complete first tracked exercise"
git pull
git push
git status
"""
        ),
        p(
            "Open the repository on GitHub, verify that the files are "
            "visible and submit the following type of link:"
        ),
        code_box(
            """
https://github.com/STUDENT-USERNAME/SBI811-Exercises/tree/main/exercise_01
""",
            GREEN_BACKGROUND,
        ),
        subheading("Submission checklist"),
        bullet_list(
            [
                "The repository is accessible and correctly named.",
                "README identifies the student and course.",
                ".gitignore is present.",
                "All questions have been answered.",
                "The FASTA file is correctly formatted.",
                "The exercise has been committed and pushed.",
                "The submitted URL opens exercise_01.",
            ]
        ),
        heading("13. Routine to remember"),
        code_box(
            """
git pull
git status
# Create or modify files
git status
git add filename
git commit -m "Describe the completed change"
git push
""",
            GREEN_BACKGROUND,
        ),
    ]
)


# ---------------------------------------------------------------------
# Build the PDF
# ---------------------------------------------------------------------

document.build(
    story,
    onFirstPage=footer,
    onLaterPages=footer,
)

print(f"Created: {OUTPUT_FILE}")
```
