# Part 5 — Values, variables and assignment

## 5.1 What is a value?

A value is a piece of information used by a program. Examples include:

- the text `"ATGCGT"`;
- the whole number `6`;
- the decimal number `50.0`; and
- the logical value `True`.

## 5.2 What is a variable?

A variable is a name that refers to a value.

```python
sequence = "ATGCGT"
```

Read this statement as:

> Assign the text `ATGCGT` to the variable named `sequence`.

The equals sign in this context means **assignment**, not a mathematical equation.


````python
sequence = "ATGCGT"
print(sequence)
````

Python stores the string and associates it with the name `sequence`. When `print(sequence)` runs, Python obtains the current value associated with that name.

## 5.3 Variables can change


````python
sample_name = "sample_A"
print(sample_name)

sample_name = "sample_B"
print(sample_name)
````

The second assignment replaces the previous value associated with `sample_name`. This is why it is called a variable: the referenced value can vary.

The old text has not been renamed; the variable now refers to a different value.


## 5.4 Choosing variable names

Good variable names make code easier to understand.

| Avoid | Prefer |
|---|---|
| `x` | `sequence_length` |
| `s` | `dna_sequence` |
| `n` | `sample_count` |
| `a1` | `gc_percentage` |

### Rules

- A name may contain letters, digits and underscores.
- A name cannot begin with a digit.
- A name cannot contain spaces.
- A name cannot contain a hyphen.
- Python keywords such as `for`, `if` and `class` cannot be variable names.
- Names are case-sensitive.

### Valid names

```text
sequence
sequence_1
sample_name
gc_percentage
```

### Invalid or problematic names

```text
1sequence       begins with a digit
sample name     contains a space
gc-content      contains a hyphen
print           hides a useful built-in function
```

## 8.4 Uppercase and lowercase

Python considers lowercase and uppercase characters different.

```text
atgc is not the same string as ATGC
```

String methods can create standardized versions.


````python
raw_sequence = "atGcgtAa"
upper_sequence = raw_sequence.upper()
lower_sequence = raw_sequence.lower()

print("Original:", raw_sequence)
print("Uppercase:", upper_sequence)
print("Lowercase:", lower_sequence)
````

The dot in `raw_sequence.upper()` means: use the `upper` method belonging to this string.

Strings are **immutable**: `upper()` creates a new string; it does not silently alter the original string.


## 8.5 Removing surrounding whitespace

The `strip()` method removes whitespace from the beginning and end of a string.


````python
raw_sequence = "   ATGCGTAA\n"
clean_sequence = raw_sequence.strip()

print("Raw representation:", repr(raw_sequence))
print("Clean representation:", repr(clean_sequence))

print("Raw representation:", raw_sequence)
print("Clean representation:", clean_sequence)

````

`repr()` displays otherwise invisible characters clearly. `\n` represents a newline.

`strip()` does not remove spaces in the middle of a string. It removes surrounding whitespace only.


## 8.6 Counting characters

Negative indices count from the end:


````python
dna = "ATGCGT"

print("Last nucleotide:", dna[-1])
print("Second-last nucleotide:", dna[-2])
````

# Part 10 — Slicing: obtaining several characters

A slice extracts part of a string.

```python
sequence[start:stop]
```

The start position is included. The stop position is excluded.

This can be remembered as:

> Start here, but stop before there.


````python
dna = "ATGCGTAA"

print(dna[0:3])
print(dna[3:6])
````

````python
dna = "ATGCGTAAGACATTCAT"
print('step 1:',dna[0:7:1])
print('step 2:',dna[0:7:2])
print('step 3:',dna[0:7:3])
````

For `dna[0:3]`, Python returns positions 0, 1 and 2. Position 3 is not included.

## 10.1 Omitting the start or stop


````python
dna = "ATGCGTAA"

print("First three:", dna[:3])
print("From position 3 onward:", dna[3:])
print("Last three:", dna[-3:])
````

## 10.2 Slice step

A third value controls the step:

```python
sequence[start:stop:step]
```

A step of 2 selects every second character.


````python
dna = "ATGCGTAA"

print("Every second character:", dna[::2])
print("Reversed string:", dna[::-1])
print(dna[:])
print(dna[0:-1:2])
````

`dna[::-1]` reverses the order of characters. It does **not** calculate a biological reverse complement because A has not been exchanged with T and G has not been exchanged with C.

# Part 13 — Biological application: GC percentage

GC percentage describes the proportion of nucleotides that are G or C.

For a sequence of length $L$:

$$
\text{GC percentage} = \frac{\text{number of G} + \text{number of C}}{L} \times 100
$$

In plain language:

1. count G;
2. count C;
3. add the two counts;
4. divide by the total sequence length; and
5. multiply by 100.


````python
dna = "ATGCGTAA"

sequence_length = len(dna)
g_count = dna.count("G")
c_count = dna.count("C")
gc_count = g_count + c_count
gc_percentage = (gc_count / sequence_length) * 100

print("Sequence:", dna)
print("Length:", sequence_length)
print("G count:", g_count)
print("C count:", c_count)
print(f"GC percentage: {gc_percentage:.2f}%")
````

## 13.1 Trace the values

For `ATGCGTAA`:

| Variable | Value | Meaning |
|---|---:|---|
| `sequence_length` | 8 | total characters/nucleotides |
| `g_count` | 2 | number of G characters |
| `c_count` | 1 | number of C characters |
| `gc_count` | 3 | G plus C |
| `gc_percentage` | 37.5 | $3/8 \times 100$ |

Tracing values step by step is a valuable debugging technique.


## 13.2 Clean before calculating

Real copied sequences may contain lowercase letters or surrounding whitespace. Standardize them before counting.


````python
raw_dna = "  atGCGtaa\n"
dna = raw_dna.strip().upper()

sequence_length = len(dna)
gc_count = dna.count("G") + dna.count("C")
gc_percentage = (gc_count / sequence_length) * 100

print("Clean sequence:", dna)
print("Length:", sequence_length)
print(f"GC percentage: {gc_percentage:.2f}%")
````

## 15.4 `IndexError`

An `IndexError` occurs when the requested position does not exist.


````python
short_sequence = "ATGC"

try:
    print(short_sequence[10])
except IndexError as error:
    print(type(error).__name__ + ":", error)
````

## 15.5 `ZeroDivisionError`

GC percentage requires division by sequence length. An empty sequence has length zero.


````python
5/0
````

````python
empty_sequence = ""

try:
    gc_percentage = (
        (empty_sequence.count("G") + empty_sequence.count("C"))
        / len(empty_sequence)
        * 100
    )
except ZeroDivisionError as error:
    print(type(error).__name__ + ":", error)
````
