# Part 1 — From calculations to decisions

Lecture 3 calculated GC percentage:

```python
gc_percentage = (gc_count / sequence_length) * 100
```

But a useful program may need to decide what the result means:

```text
If GC percentage is at least 60, report high GC.
Otherwise, report moderate or low GC.
```

This requires a **condition**: an expression that Python evaluates as either `True` or `False`.


# Part 2 — Boolean values and comparisons

## 2.1 Boolean values

The Boolean type has only two possible values:

```python
True
False
```

The capital letters matter. `true` and `false` are not Python Boolean values.


````python
passed_quality_check = True
contains_unknown_bases = False

print(passed_quality_check)
print(contains_unknown_bases)
print(type(passed_quality_check))
````

## 2.2 Comparison operators

Comparison operators compare values and return `True` or `False`.

| Operator | Meaning | Example |
|---|---|---|
| `==` | equal to | `length == 100` |
| `!=` | not equal to | `base != "N"` |
| `>` | greater than | `gc > 60` |
| `<` | less than | `length < 50` |
| `>=` | greater than or equal to | `quality >= 30` |
| `<=` | less than or equal to | `n_count <= 2` |


````python
sequence_length = 24

print(sequence_length == 24)
print(sequence_length != 24)
print(sequence_length > 20)
print(sequence_length < 20)
print(sequence_length >= 24)
print(sequence_length <= 23)
````

## 2.3 Assignment and comparison are different

```python
sequence_length = 24
```

The single equals sign **assigns** the value 24.

```python
sequence_length == 24
```

The double equals sign **compares** the current value with 24.

Read `==` as “is equal to?”

<div class="warning-box">
Using <code>=</code> where Python expects a comparison is a very common beginner mistake.
</div>

The `in` operator asks whether a value occurs inside another value or collection.


````python
dna = "ATGCGTAA"

print("ATG" in dna)
print("NNN" in dna)
print("N" not in dna)
````

# Part 3 — Combining conditions

## 3.1 The `and` operator

`and` produces `True` only when **both** conditions are true.

Example biological rule:

> A sequence passes when its length is at least 20 and it contains no N.


````python
dna = "ATGCGTAACTGGCTAACGTA"

length_is_sufficient = len(dna) >= 20
has_no_unknown_bases = "N" not in dna

passes = length_is_sufficient and has_no_unknown_bases

print("Length sufficient:", length_is_sufficient)
print("No unknown bases:", has_no_unknown_bases)
print("Passes both checks:", passes)
````

## 3.2 The `or` operator

`or` produces `True` when **at least one** condition is true.

Example rule:

> A nucleotide is a purine when it is A or G.


````python
base = "G"
is_purine = base == "A" or base == "G"

print(is_purine)
````

The membership form is shorter and often clearer:

```python
is_purine = base in "AG"
```


````python
base = "G"
is_purine = base in "AG"

print(is_purine)
````

## 3.3 The `not` operator

`not` reverses a Boolean result.


````python
contains_n = "N" in "ATGCNTA"

print("Contains N:", contains_n)
print("Does not contain N:", not contains_n)
````

## 3.4 Truth tables in plain language

| A | B | `A and B` | `A or B` |
|---|---|---|---|
| True | True | True | True |
| True | False | False | True |
| False | True | False | True |
| False | False | False | False |

- `and` is strict: every required condition must pass.
- `or` is flexible: one passing condition is enough.
- `not` reverses the answer.

# Part 4 — Making decisions with `if`

## 4.1 A simple `if` statement

```python
if condition:
    instruction
```

Python executes the indented instruction only when the condition is `True`.


````python
dna = "ATGCGTAA"

if "ATG" in dna:
    print("Start motif found")
    
````

Important features:

1. `if` begins the decision.
2. The condition follows `if`.
3. A colon `:` ends the `if` line.
4. The controlled instruction is indented.

If the condition is false, the indented block is skipped.


````python
dna = "CCGCGTAA"

if "ATG" in dna:
    print("Start motif found")
    
print("Sequence check completed")    


````

`Sequence check completed` is not indented, so it runs regardless of the condition.


## 4.2 Indentation defines a block

Python commonly uses four spaces for each indentation level.

Correct:

```python
if gc_percentage >= 50:
    print("GC is at least 50%")
```

Incorrect:

```python
if gc_percentage >= 50:
print("GC is at least 50%")
```

The second example causes an `IndentationError` because Python expects an indented block.


````python
if gc_percentage >= 50:
print("GC is at least 50%")
````

## 4.3 `if` and `else`

`else` provides an alternative when the condition is false.


````python
dna = "ATGCNTAA"

if "N" not in dna:
    print("Sequence contains only identified bases")
else:
    print("Sequence contains at least one unknown base")
````

Exactly one of these two blocks is executed.


## 4.4 More alternatives with `elif`

`elif` means “else if”. It checks another condition when previous conditions were false.


````python
gc_percentage = 57.5

if gc_percentage >= 60:
    print("High GC")
elif gc_percentage >= 40:
    print("Moderate GC")
else:
    print("Low GC")
````

Python tests the branches from top to bottom and stops after the first true condition.

For `57.5`:

1. `57.5 >= 60` is false;
2. `57.5 >= 40` is true;
3. Python prints `Moderate GC`;
4. the `else` block is skipped.


## 4.5 The order of conditions matters

This order is incorrect:

```python
if gc_percentage >= 40:
    print("Moderate or high")
elif gc_percentage >= 60:
    print("High")
```

A value of 70 satisfies the first condition, so Python never reaches the more specific high-GC condition.

When thresholds overlap, test the more restrictive condition first.


````python
gc_percentage = 70

if gc_percentage >= 40:
    print("Moderate or high")
elif gc_percentage >= 60:
    print("High")
    
else:
    print('less GC')
````

## 4.6 Nested conditions

An `if` block may contain another `if`. This is called nesting.


````python
dna = "ATGCGTAA"

if len(dna) > 0:
    print("Sequence is not empty")

    if "N" in dna:
        print("Unknown bases detected")
    else:
        print("No unknown bases detected")
else:
    print("Sequence is empty")
````

Nesting is useful, but many levels become difficult to read. Use it only when the second decision logically depends on the first.

## 7.4 Slicing a list

List slices follow the same start-included, stop-excluded rule as string slices.


````python
sample_names = ["sample_A", "sample_B", "sample_C", "sample_D"]

print(sample_names[:2])
print(sample_names[1:3])
print(sample_names[-2:])
````

The result of slicing a list is another list.


## 7.5 Lists are mutable

**Mutable** means that an object can be changed after it is created.

Replace an element by assigning a new value at its index.


````python
sample_names = ["sample_A", "sample_B", "sample_C"]
print("Before:", sample_names)

sample_names[1] = "sample_B_corrected"
print("After:", sample_names)
````

````python
sample_names = ["sample_A", "sample_B", "sample_C"]

print(sample_names)

sample_names[2] = 'sample_D'

print(sample_names)
````

The list itself was changed. This is different from strings, which are immutable.


## 7.6 Adding items with `append()`

`append()` adds one item to the end of a list.


````python
sequences = ["ATGC", "GCTA"]
print("Before:", sequences)

sequences.append("TTAA")
print("After:", sequences)
````

````python
sequences = ["ATGC", "GCTA"]
sequences.append([1,8])

print(sequences)
````

`append()` modifies the existing list. It does not create a separate returned list.

A common mistake is:

```python
sequences = sequences.append("TTAA")
```

After this statement, `sequences` becomes `None` because `append()` changes the list in place and returns no new list.


````python
demonstration = ["ATGC", "GCTA"]
returned_value = demonstration.append("TTAA")

print("Modified list:", demonstration)
print("Value returned by append:", returned_value)
````

## 7.7 Inserting at a specific position

`insert(index, value)` adds an item at a selected position.


````python
sample_names = ["sample_A", "sample_C"]
sample_names.insert(1, "sample_B")

print(sample_names)
````

Existing items at and after the insertion position move to the right.


## 7.8 Removing items

`remove(value)` removes the first matching value.


````python
sample_names = ["sample_A", "failed_sample", "sample_C"]
sample_names.remove("failed_sample")

print(sample_names)
````

`pop(index)` removes and returns an item at an index. Without an index, it removes the last item.


````python
sample_names = ["sample_A", "sample_B", "sample_C"]
removed_sample = sample_names.pop(1)

print("Removed:", removed_sample)
print("Remaining:", sample_names)
````

Use `remove()` when you know the value. Use `pop()` when you know the position and may need the removed value.

# Part 8 — Tuples

## 8.1 Creating a tuple

A tuple commonly uses parentheses.


````python
allowed_bases = ("A", "T", "G", "C")

print(allowed_bases)
print(type(allowed_bases))
````

Tuples support indexing, slicing, `len()` and membership tests.

## 8.2 Tuples are immutable

Immutable means that items cannot be replaced, appended or removed after the tuple is created.

The following would cause a `TypeError`:

```python
allowed_bases[0] = "U"
```

A tuple is useful when a collection represents fixed information that should not change accidentally.


## 8.3 List versus tuple

| Feature | List | Tuple |
|---|---|---|
| Syntax | `["A", "T"]` | `("A", "T")` |
| Can replace items? | Yes | No |
| Can append items? | Yes | No |
| Typical use | Changing collection | Fixed collection |

For example:

- a growing list of sample names can be a list;
- the fixed set of four expected DNA bases can be a tuple.


## 8.4 A one-item tuple needs a comma

```python
not_a_tuple = ("A")

```


````python
not_a_tuple = ("A")
one_item_tuple = ("A",)

print(type(not_a_tuple))
print(type(one_item_tuple))
````

The comma, not the parentheses alone, creates the one-item tuple.

# Part 10 — `for` loops

## 10.1 Basic structure

```python
for item in collection:
    instruction_using_item
```

Read it as:

> For each item in the collection, run the indented instruction.


````python
sequences = ["ATGC", "GCTAAG", "TTAA"]

for sequence in sequences:
    print(sequence)
````

````python
dna_seq = ["ATGC", "GCTAAG", "TTAA"]

for i in dna_seq:
    print(i)
````

````python
sequences = ["ATGC", "GCTAAG", "TTAA"]

print(sequences[0])
print(sequences[1])
print(sequences[2])
````

Step by step:

1. `sequence` refers to `"ATGC"`; the body runs.
2. `sequence` refers to `"GCTAAG"`; the body runs.
3. `sequence` refers to `"TTAA"`; the body runs.
4. No items remain; the loop ends.

The singular loop-variable name `sequence` and plural list name `sequences` help communicate their different roles.


## 10.2 Process each sequence


````python
sequences = ["ATGC", "GCTAAG", "TTAA"]

for seq in sequences:
    length = len(seq)
    print(seq, "has length", length)
````

The two indented statements belong to the loop and therefore run for every sequence.


## 10.3 Code after a loop


````python
sequences = ["ATGC", "GCTAAG", "TTAA"]

for sequence in sequences:
    print("Processing:", sequence)

print("All sequences processed")
````

The final line is not indented, so it runs once after the loop finishes.


## 10.4 Loop over a string

A string is a sequence of characters, so a `for` loop can visit each character.


````python
dna = "ATGC"

for base in dna:
    print(base)
````

This provides a complete way to validate every base.


````python
dna = "ATGCXTA"
allowed_bases = ("A", "T", "G", "C")

for base in dna:
    if base in allowed_bases:
        print(base, "is valid")
    else:
        print(base, "is invalid")
````

The `if` statement is inside the loop, so each base is checked separately.


## 10.5 Building a count with an accumulator

An accumulator is a variable updated repeatedly to build a result.


````python
dna = "ATGCGTAA"
g_count = 0

for base in dna:
    if base == "G":
        g_count = g_count + 1

print("G count:", g_count)
````

````python
dna = "ATGCGTAATTGCTTTC"

t_count = 0

for i in dna:
    
    if i == 'T':
        
        t_count = t_count + 1
        
print(t_count)
````

````python
dna = "ATGCGTAATTGCTTTC"

t_count = 0

for nuc in dna:
    if nuc == 'T':
        t_count += 1
        
print(t_count)        
    
    
````

Trace the value:

```text
Start: g_count = 0
See A: unchanged
See T: unchanged
See G: g_count becomes 1
See C: unchanged
See G: g_count becomes 2
...
```

The shorter update `g_count += 1` means the same as `g_count = g_count + 1`.


````python
dna = "ATGCGTAA"
g_count = 0

for base in dna:
    if base == "G":
        g_count += 1

print("G count:", g_count)
````

# Part 11 — `range()`

`range()` generates a sequence of integers, commonly used with a `for` loop.

## 11.1 `range(stop)`

The stop value is excluded.


````python
for number in range(5):
    print(number)
````

The values are 0, 1, 2, 3 and 4.

The step of 2 generates 0, 2, 4, 6 and 8.

# Part 12 — `enumerate()`

`enumerate()` provides both an index and the corresponding item.


````python
dna = "ATGCGT"

for index, base in enumerate(dna):
    print(index, base)
````

This is often clearer than manually indexing with `range(len(dna))`.

To display positions beginning at 1 for a human reader:


````python
dna = "ATGCGT"

for position, base in enumerate(dna, start=1):
    print(position, base)
````

The biological display position begins at 1, although the underlying Python string remains zero-indexed.

# Part 13 — `while` loops

A `for` loop is appropriate when processing items in a collection. A `while` loop repeats while a condition remains true.

```python
while condition:
    repeated_instruction
```


````python
count = 1

while count <= 5:
    print(count)
    count += 1
````

The update `count += 1` is essential. Without it, `count <= 5` remains true and the loop may continue indefinitely.

<div class="warning-box">
If a notebook cell runs indefinitely, interrupt the kernel. In Jupyter, select <b>Kernel → Interrupt Kernel</b> or use the stop button.
</div>


## 13.1 Trace a `while` loop

```text
count = 1 → condition true → print 1 → count becomes 2
count = 2 → condition true → print 2 → count becomes 3
...
count = 5 → condition true → print 5 → count becomes 6
count = 6 → condition false → loop stops
```


## 13.2 A controlled trimming example

Remove leading `N` characters one at a time.


````python
dna = "NNNATGCGT"

while dna.startswith("N"):
    dna = dna[1:]

print("Trimmed sequence:", dna)
````

Each repetition removes the first character. The loop stops when the sequence no longer begins with `N`.

In normal code, `lstrip("N")` could perform this particular task more directly. The loop is shown to demonstrate repeated condition-controlled change.


# Part 14 — `break` and `continue`

These statements alter normal loop flow. Use them sparingly and only when they make the logic clearer.

## 14.1 `break` stops a loop


````python
dna = "ATGCXTAA"
allowed_bases = ("A", "T", "G", "C")

for position, base in enumerate(dna, start=1):
    if base not in allowed_bases:
        print("Invalid base", base, "at position", position)
        break
````

After the first invalid base is found, no further positions are checked.

## 14.2 `continue` skips the remainder of one iteration


````python
dna = "ATGCNTAA"

for base in dna:
    if base == "N":
        continue
    print("Identified base:", base)
````

When the base is `N`, Python skips the print statement and moves to the next character.

````python
# determine GC percentage

sequences = ["ATGCGTAA", "ggccaatt", "ATGCNNTA", ""]

for raw_sequence in sequences:
    sequence = raw_sequence.strip().upper()

    print("Sequence:", repr(sequence))

    if len(sequence) == 0:
        print("Status: empty sequence")
    elif "N" in sequence:
        print("Status: contains unknown bases")
    else:
        length = len(sequence)
        gc_count = sequence.count("G") + sequence.count("C")
        gc_percentage = (gc_count / length) * 100

        print("Length:", length)
        print(f"GC percentage: {gc_percentage:.2f}%")
````

## 15.1 Why the empty check comes first

An empty sequence has length zero. Dividing by zero would cause a `ZeroDivisionError`. The first condition prevents the calculation for empty input.

## 15.2 Why the unknown-base check comes second

The program applies a stated rule: basic GC percentage is reported only for sequences without `N`. Another analysis might use a different policy. Good scientific code makes such decisions explicit.
