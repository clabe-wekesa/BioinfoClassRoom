## 1.1 The problem with repeated code

Suppose we must calculate GC percentage for three sequences. We could copy the calculation three times, but copied code is longer and easier to make inconsistent.

````python
sequence_1 = "ATGCGC"
gc_1 = sequence_1.count("G") + sequence_1.count("C")
percentage_1 = (gc_1 / len(sequence_1)) * 100

sequence_2 = "AAAAAT"
gc_2 = sequence_2.count("G") + sequence_2.count("C")
percentage_2 = (gc_2 / len(sequence_2)) * 100

sequence_3 = "GGCCGG"
gc_3 = sequence_3.count("G") + sequence_3.count("C")
percentage_3 = (gc_3 / len(sequence_3)) * 100

print(percentage_1, percentage_2, percentage_3)
````

A function lets us write the calculation once and reuse it. Benefits include:

- **maintenance** — correct the logic in one place; and
- **testing** — check one small task independently.

# Part 4 — `print()` and `return` are different

This distinction is essential.

- `print()` displays a value for a person to see.
- `return` sends a value back to the part of the program that called the function.

A printed value is not automatically available for a later calculation.

````python
def print_length(sequence):
    print(len(sequence))

printed_result = print_length("ATGCGT")
print("Stored value:", printed_result)
````

The function displayed `6`, but `printed_result` contains `None`. `None` means no useful value was returned.

````python
def get_length(sequence):
    return len(sequence)

returned_result = get_length("ATGCGT")
print("Stored value:", returned_result)
print("Twice the length:", returned_result * 2)
````

`return` makes the result reusable. We can store it, compare it, print it or pass it to another function.

````python
length = get_length("ATGCGT")

if length >= 6:
    print("The sequence meets the minimum length.")
````

## 5.3 Calculate GC percentage

The formula is:

$$GC\% = \frac{G + C}{\text{sequence length}} \times 100$$

We must decide what should happen for an empty sequence because division by zero is impossible.

````python
def calculate_gc_percentage(sequence):
    cleaned_sequence = sequence.strip().upper()

    if len(cleaned_sequence) == 0:
        return 0.0

    gc_count = cleaned_sequence.count("G") + cleaned_sequence.count("C")
    gc_percentage = (gc_count / len(cleaned_sequence)) * 100
    return gc_percentage

print(calculate_gc_percentage("ATGCGC"))
print(calculate_gc_percentage("AAAAAT"))
print(calculate_gc_percentage(""))
````

````python
def calculate_gc_percentage(sequence):
    
    gc_count = 0
    
    for base in sequence.strip().upper():
        if base == 'G' or base == 'C':
            gc_count = gc_count + 1
    
    gc_percent = gc_count/len(sequence) * 100
    return gc_percent   
                
calculate_gc_percentage(sequence='AAAAAT')   
````

Returning `0.0` for an empty sequence is a design decision suitable for this beginner exercise. In scientific software, the team should document such decisions clearly.

# Part 6 — Functions can call other functions

We do not need to repeat cleaning logic in every function. One function may call another.

````python
def clean_sequence(sequence):
    cleaned = sequence.strip().upper()
    return cleaned

def calculate_gc_percentage_v2(sequence):
    cleaned_sequence = clean_sequence(sequence)
    if len(cleaned_sequence) == 0:
        return 0.0
    gc_count = cleaned_sequence.count("G") + cleaned_sequence.count("C")
    return (gc_count / len(cleaned_sequence)) * 100

print(calculate_gc_percentage_v2("  atgcgc\n"))
````

This creates a simple flow:

```text
raw sequence → clean_sequence() → calculate_gc_percentage_v2() → numeric result
```

Small functions can become building blocks for a larger analysis.

## 6.1 Validate DNA

For this lecture, a valid DNA sequence contains only `A`, `T`, `G` and `C`. We will return `True` or `False`.

````python
def clean_sequence(sequence):
    cleaned = sequence.strip().upper()
    return cleaned

def is_valid_dna(sequence):
    cleaned_sequence = clean_sequence(sequence)

    if len(cleaned_sequence) == 0:
        return False

    for base in cleaned_sequence:
        if base not in "ATGC":
            return False

    return True

print(is_valid_dna("ATGCGT"))
print(is_valid_dna("ATGXGT"))
print(is_valid_dna(""))
````

Notice the placement of the final `return True`. Python can report success only after the loop has checked **every** base.

## 6.2 Classify GC percentage

````python
def classify_gc(sequence):
    gc_percentage = calculate_gc_percentage_v2(sequence)

    if gc_percentage >= 60:
        return "high GC"
    elif gc_percentage >= 40:
        return "moderate GC"
    else:
        return "low GC"

print(classify_gc("GGCCGG"))
print(classify_gc("ATGCAT"))
print(classify_gc("AAAAAT"))
````

# Part 7 — Multiple and default parameters

A function may receive more than one argument.

````python
def sequence_meets_length(sequence, minimum_length):
    cleaned_sequence = clean_sequence(sequence)
    return len(cleaned_sequence) >= minimum_length

print(sequence_meets_length("ATGCGT", 5))
print(sequence_meets_length("ATG", 5))
````

Arguments are normally matched to parameters by position. In `sequence_meets_length("ATG", 5)`, `"ATG"` becomes `sequence`, and `5` becomes `minimum_length`.

## 7.1 Default parameter values

A default supplies a value when the caller does not provide one.

````python
def sequence_meets_length_v2(sequence, minimum_length=5):
    cleaned_sequence = clean_sequence(sequence)
    return len(cleaned_sequence) >= minimum_length

print(sequence_meets_length_v2("ATGCGT"))      # uses 5
print(sequence_meets_length_v2("ATGCGT", 10))  # uses 10
````

Use defaults only when there is a sensible, clearly documented usual value. Required parameters come before default parameters.

# Part 8 — Local variables and scope

A variable created inside a function is normally **local** to that function. It exists while the function runs and helps prevent unrelated parts of a program from interfering with one another.

````python
def calculate_length(sequence):
    sequence_length = len(sequence)
    return sequence_length

result = calculate_length("ATGC")
print(result)
````

`sequence_length` is local. Trying to use it outside the function would cause `NameError`. The returned value is stored outside as `result`.

Avoid relying on global variables when a function can receive an argument instead.

````python
dna_sequence = "ATGC"

def unclear_length():
    return len(dna_sequence)  # depends on an outside variable

def clear_length(sequence):
    return len(sequence)      # receives its dependency explicitly

print(unclear_length())
print(clear_length("ATGC"))
````

# Part 9 — Docstrings: explaining a function

A **docstring** is a short description placed immediately inside a function using triple quotes. It should state what the function does, not narrate every line.

````python
def calculate_gc_with_docstring(sequence):
    """Return the GC percentage of a DNA sequence."""
    cleaned_sequence = clean_sequence(sequence)
    if len(cleaned_sequence) == 0:
        return 0.0
    gc_count = cleaned_sequence.count("G") + cleaned_sequence.count("C")
    return (gc_count / len(cleaned_sequence)) * 100

help(calculate_gc_with_docstring)
````

For beginner functions, one clear sentence is often enough. A longer docstring may describe parameters, the returned value and special cases.

````python
def count_nucleotide_documented(sequence, nucleotide):
    """Count one nucleotide in a DNA sequence.

    sequence: DNA text to examine.
    nucleotide: one base such as A, T, G or C.
    Returns: an integer count.
    """
    cleaned_sequence = clean_sequence(sequence)
    return cleaned_sequence.count(nucleotide.upper())

print(count_nucleotide_documented("ATGATA", "a"))
````

# Part 10 — Reverse complement as a function

DNA strands are antiparallel. A useful operation is to:

1. replace each base with its complement (`A↔T`, `G↔C`); then
2. reverse the complemented sequence.

We will use conditions and a loop, avoiding dictionaries until a later lesson.

````python
def reverse_complement(sequence):
    """Return the reverse complement of a valid DNA sequence."""
    cleaned_sequence = clean_sequence(sequence)
    complement = ""

    for base in cleaned_sequence:
        if base == "A":
            complement = complement + "T"
        elif base == "T":
            complement = complement + "A"
        elif base == "G":
            complement = complement + "C"
        elif base == "C":
            complement = complement + "G"
        else:
            return "Invalid DNA sequence"

    return complement[::-1]

print(reverse_complement("ATGC"))
print(reverse_complement("atgc"))
print(reverse_complement("ATGX"))
````

For `ATGC`:

- complement: `TACG`
- reverse complement: `GCAT`

The slice `[::-1]` makes a reversed copy of the string.

# Part 11 — Debugging: finding and correcting errors

**Debugging** is the systematic process of finding and correcting a problem. An error is evidence about what Python could not do; it is not evidence that you cannot program.

Use this routine:

1. read the final line of the traceback;
2. identify the error type and message;
3. find the referenced line;
4. inspect names, types, punctuation and indentation;
5. make one small correction; and
6. run the smallest relevant test again.

## 11.1 Reading a traceback

A traceback is read mostly from the bottom. For example:

```text
NameError: name 'sequnce' is not defined
```

The type is `NameError`; the message points to a misspelled name. Error demonstrations below are caught so this teaching notebook can still run from top to bottom.

````python
try:
    print(sequnce)
except NameError as error:
    print("Caught error:", type(error).__name__)
    print("Message:", error)
````

## 11.2 Common error types

| Error | Usual meaning | First thing to inspect |
|---|---|---|
| `NameError` | A name does not exist | spelling; whether a defining cell ran |
| `TypeError` | Operation used incompatible types | value types and function arguments |
| `ZeroDivisionError` | Division by zero | empty data or zero denominator |
| `SyntaxError` | Python cannot parse the code | colon, brackets, quotes, spelling |
| `IndentationError` | Indentation is inconsistent or missing | spaces below `def`, `if`, or loop |
| `AssertionError` | A test expectation was not met | function result or expected value |

### `TypeError` example

````python
try:
    print("Length: " + 6)
except TypeError as error:
    print("Caught error:", type(error).__name__)
    print("Message:", error)

# Two correct alternatives:
print("Length:", 6)
print("Length: " + str(6))
````

### `ZeroDivisionError` example

````python
def unsafe_gc(sequence):
    gc_count = sequence.count("G") + sequence.count("C")
    return (gc_count / len(sequence)) * 100

try:
    unsafe_gc("")
except ZeroDivisionError as error:
    print("Caught error:", type(error).__name__)
    print("Message:", error)
````

The corrected function checks for an empty sequence before division, as `calculate_gc_percentage()` already does.

# Part 13 — Integrated DNA-analysis workflow

We now combine small, tested functions to analyse several short sequences. The workflow will:

1. clean each sequence;
2. validate it;
3. calculate length and GC percentage;
4. classify its GC content; and
5. calculate its reverse complement.

````python
sample_sequences = [
    " atgcgc\n",
    "AAAAAT",
    "GGCCGG",
    "ATGXTA"
]

for raw_sequence in sample_sequences:
    sequence = clean_sequence(raw_sequence)
    print("Sequence:", sequence)

    if is_valid_dna(sequence):
        print("Valid DNA: Yes")
        print("Length:", get_length(sequence))
        print("GC percentage:", round(calculate_gc_percentage_v2(sequence), 2))
        print("GC class:", classify_gc(sequence))
        print("Reverse complement:", reverse_complement(sequence))
    else:
        print("Valid DNA: No — analysis skipped")

    print("-" * 35)
````

## Why this design is easier to trust

The loop coordinates the workflow, while each function has one job:

| Function | Responsibility |
|---|---|
| `clean_sequence()` | standardize sequence text |
| `is_valid_dna()` | accept or reject DNA symbols |
| `get_length()` | return sequence length |
| `calculate_gc_percentage_v2()` | calculate GC percentage |
| `classify_gc()` | interpret GC percentage |
| `reverse_complement()` | construct reverse complement |

If a reverse complement is wrong, we can examine that function without searching through one very large block.

````python
def sequence_report(sequence, minimum_length=5):
    """Return a short report for a DNA sequence."""
    cleaned_sequence = clean_sequence(sequence)

    if not is_valid_dna(cleaned_sequence):
        return "Invalid DNA sequence"

    length = len(cleaned_sequence)
    gc_percentage = calculate_gc_percentage_v2(cleaned_sequence)
    meets_minimum = length >= minimum_length
    reverse_comp = reverse_complement(cleaned_sequence)

    return (
        f"Length={length}; GC={gc_percentage:.1f}%; "
        f"minimum length met={meets_minimum}; "
        f"reverse complement={reverse_comp}"
    )

print(sequence_report("ATGCAT"))
print(sequence_report("ATGXAT"))
````
